# Tier 2: the width by weight-decay grid and the (a, b) fit

## What this notebook found

All 100 runs of the grid are in: four widths, five decay levels and five seeds, every run in NTK
parameterisation at alpha 1. The generalisation event is the first step at test accuracy 1.0. Of the 100
runs, 95 memorised and can enter the fits. The other five are the five seeds at N = 1600 with decay 0.0003,
so that cell has no grokking time at all. Of the 95 usable runs, 12 are censored: all ten no-decay runs at
N = 400 and N = 1600, and two runs at N = 50.

**The head claim: both exponents are nonzero, so neither single clock fits.** The main-effects fit over the
75 nonzero-decay runs, one of them censored, gives a width exponent a = +0.31 [+0.23, +0.38] and a decay
exponent b = -0.49 [-0.56, -0.42]. H predicts a > 0 with b near 0. The weight-decay clock predicts a near 0
with b near -1. The bootstrap cloud in section 5 excludes both. In the report's terms, two nonzero exponents
mean two clocks. Measured from step 0, which is how Kim (2026) measures, the exponents are
a = +0.15 [+0.10, +0.20] and b = -0.38 [-0.42, -0.33]. That is further from Kim's -1, not closer. The
direction check passes, because a is positive, so there is no sign of a parameterisation fault.

**The surface is not a plane, and the interaction is the main result.** The interaction coefficient is
-0.21 [-0.27, -0.15], and adding it lowers the residual sigma from 0.60 to 0.52 on the log scale. Fitted
within each decay level, the width exponent falls from +0.61 at 1e-5 to +0.32 at 3e-5, +0.16 at 1e-4 and
-0.21 at 3e-4. Fitted within each width, the decay exponent is -0.15 at N = 50, with an interval that
includes zero, -0.38 at N = 100, -0.82 at N = 400 and -0.71 at N = 1600. So at the narrow widths decay barely
changes the grokking time. At the wide widths the grokking time falls almost in proportion to 1/(eta lambda),
which is close to Kim's prediction. Section 6 flags four cells whose mean residual exceeds twice its standard
error, and they are the ones this pattern predicts. N = 50 at 1e-4 and 3e-4 and N = 400 at 1e-5 grok more
slowly than the plane predicts, and N = 100 at 1e-5 groks faster.

**Without decay, the grokking time grows steeply with width, and by N = 400 the runs do not grok within
300,000 steps.** At N = 50 and N = 100 the no-decay runs grok, with a median of about 30,000 steps. All ten
no-decay runs at N = 400 and N = 1600 are still censored at 300,000 steps. The reciprocal fit puts the
no-decay grokking rate at 0.55, 0.44, 0.22 and -0.01 [-0.05, +0.04] per 10,000 steps at N = 50, 100, 400 and
1600. The rate at N = 1600 comes from a fit that is linear in log N, and there it rests on censored runs
only. A pure weight-decay clock requires c0 = c2 = 0, and the joint test rejects that, with a Wald statistic
of 89.8 and p = 3e-20. At N = 100 decay accounts for 9 percent of the fitted rate at 1e-5 and 75 percent at
3e-4.

**One reading, stated as a hypothesis and not as a result.** The pattern fits two routes to generalisation
that run side by side, with the faster one setting the time. One route runs without decay and slows sharply
with width, and it dominates at the narrow widths. The other runs on decay, close to Kim's 1/(eta lambda), and
it takes over once the network is wide enough that the first route has stalled. The reciprocal form adds the
two rates, which is itself a model of routes running side by side. The report instead says that with two
clocks "the slower one sets the time", and these data point the other way. A direct comparison of the two
forms has not been fitted.

**The alignment threshold roughly survives across width, but drifts upward.** Across the 83 runs that
generalised, the alignment at the event has a max-to-min ratio of 1.75 and a spread of 11.7 percent, so it
stays inside the kill criterion of about 2. Its mean rises with width, however: 0.107, 0.121, 0.132 and 0.131
from N = 50 to N = 1600. The threshold therefore rises by about a quarter between N = 50 and N = 400 and then
levels off, so it is not strictly shared. In the placebo test the alignment spread at the true events was
smaller than at the swapped events in all 2,000 permutations. The tightening is only 1.1-fold, weaker than
the 1.6-fold of Tier 1.

**Across width, the task-blind shape term R is not locked to the event.** Its value at the event falls from
0.158 at N = 50 to 0.006 at N = 1600, a max-to-min ratio of 38, and the placebo shows no tightening. This
reverses the Tier 1 finding that R tracked the event more tightly than A did. Across width, the
task-referenced alignment is the better threshold variable.

**Crossing the alignment band is again not sufficient.** All five seeds at N = 1600 with decay 0.0003 reach
a peak alignment of 0.123 to 0.127, inside the band where other runs generalise. None of them keeps train
accuracy 1, and none generalises. Train accuracy peaks between 0.90 and 0.93 and ends between 0.68 and 0.75,
while the weight norm falls to 0.31 of its initial value. This is a second failure of sufficiency, after the
alpha 2 runs at decay 0.0003 in Tier 1.

**The early shape rate falls with width, but here it cannot be separated from width.** The median early
alignment rate over steps 0 to 2,000 is 0.0094, 0.0078, 0.0035 and 0.0012 per 1,000 steps at N = 50, 100,
400 and 1600. So wider networks do reshape more slowly, which is the premise of H. The censored slope of log
grokking time on the log early rate is -0.89 [-1.02, -0.78], and log N adds nothing once the rate is in the
model. However, the rate is almost a function of width alone, and within one width the decay moves the
grokking time tenfold with no change in the early rate. The early rate therefore carries the width effect
and none of the decay effect. The window was fixed in advance, but in 40 of the 95 runs, all at N of 400 or
more, memorisation happens inside it, at step 500. So the check the section set for itself fails at the wide
widths.

**Sensitivity.** The signs of a and b hold at test accuracy levels 1.0, 0.95 and 0.9. The width exponent
rises from +0.31 to +0.50 as the level is lowered, and b stays between -0.49 and -0.48. Under the loss
definition, 53 of the 90 usable runs are censored, so its fit cannot be read.

**What this means for the report.** In direction, H's width prediction holds. The width exponent is
positive, and strongly so without decay, and wider networks reshape more slowly. H's decay prediction,
b near 0, holds only at N = 50. The weight-decay clock's b = -1 is approached only at N of 400 or more. The
pre-registered head claim therefore comes out as two clocks, with an interaction that the report's plane
cannot represent. The report should present the within-stratum slopes and the reciprocal fit beside the
(a, b) numbers, and it should revisit the sentence that the slower clock sets the time. Every number here
comes from one task, one depth and one value of alpha. The no-decay result at the wide widths rests on lower
bounds set by the 300,000-step budget.


## Grid

Every run uses the NTK parameterisation at alpha 1 and base rate 100, with p = 23, a training fraction of 0.9
and data seed 42. The width N is 50, 100, 400 or 1600 and eta times kappa is 0, 0.00001, 0.00003, 0.0001 or
0.0003, with model seeds 0 to 4 in every cell. Runs at N = 50 and N = 100 last 100,000 steps and runs at
N = 400 and N = 1600 last 300,000 steps, with a checkpoint every 500 steps. The N = 100 cells are the Tier 1
runs at alpha 1, reloaded rather than retrained. The grid is defined once, in `tier2_sweep.py`.

Three things differ from the report. The decay axis is the Tier 1 axis rather than the report's
0.001 to 0.1, because a decay of 0.001 stops the N = 100 network from memorising. The step budget depends on
the width. The events are defined on accuracy, as in Tiers 0 and 1. The reasons for each, the source check on
Kim (2026), and every observation made along the way are recorded in `tier_2.md`.

## How to read the tables and figures

**Grokking time** is the number of steps between two events: memorisation, when the network gets the
training set right, and generalisation, when it gets the held-out test pairs right. Memorisation is the first
step at train accuracy 1. Generalisation is the first step at which test accuracy reaches `TEST_ACC_LEVEL`,
set in the first code cell. **The generalisation step** t_test is the same event counted from step 0. It is
the quantity that Kim (2026), Section 7.5, fits.

**Censoring.** If generalisation has not happened when the run ends, the grokking time is known only to
exceed the run length minus the memorisation step. Tables write such a bound with a plus sign. Figures draw
it as a triangle pointing towards the true value, and a measured time as a circle. A run that never
memorises has no grokking time at all. It is counted in the tables and left out of every fit.

**Dots on curves.** A hollow dot is the memorisation event and a filled dot the generalisation event of that
run.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, chi2
import ntk_lib as L  # training loop, crossings, kernel terms and plot helpers shared by the tier notebooks
import tier2_sweep as T  # the Tier 2 grid: widths, decays, seeds, alpha and the step budget of each width

L.set_plot_style()

WIDTHS, DECAYS, SEEDS = T.WIDTHS, T.DECAYS, T.SEEDS
NONZERO = [ek for ek in DECAYS if ek > 0]

# Test accuracy at which the generalisation event is declared; the memorisation event is always train accuracy 1.
# Change this one number to move every crossing, table and figure in the notebook.
TEST_ACC_LEVEL = 1.0

# Every saved run of the grid. Missing runs are skipped, so the notebook also works on a partial sweep.
# The runs themselves are trained by tier2_sweep.py.
grid = {(N, ek, s): T.load(N, ek, s) for N in WIDTHS for ek in DECAYS for s in SEEDS if T.is_saved(N, ek, s)}
print(f"{len(grid)} of {len(WIDTHS) * len(DECAYS) * len(SEEDS)} runs are saved.")

width_colours = dict(zip(WIDTHS, L.RAMP[1:]))  # light for narrow, dark for wide
decay_colours = dict(zip(DECAYS, L.RAMP))      # light for no decay, dark for the strongest


def records(definition="accuracy", test_level=TEST_ACC_LEVEL, tau=1e-2):
    """One dictionary per saved run with its cell, its two events, its times and its kernel terms at the event.

    A run that never memorised has no grokking time and is marked as not memorised rather than censored.
    t_test is the generalisation step counted from step 0; a run that never generalised has it censored at
    its last step, and its grokking time censored at the last step minus the memorisation step.
    """
    out = []
    for (N, ek, s), run in grid.items():
        r = L.summarise_cell(run, definition, tau, test_level)
        out.append(dict(N=N, ek=ek, seed=s, cell=f"{N}_{ek:g}", memorised=r["t_train"] is not None,
                        censored=r["censored"], t_train=r["t_train"], t_grok=r["t_grok"],
                        t_test=r["t_test"] if r["t_test"] is not None else r["last_step"],
                        A0=run["history"]["A_t"][0], A_ev=r["A_at_test"], R_ev=r["R_at_test"], S_ev=r["S_at_test"]))
    return out


def usable(recs):
    """The runs that can enter a fit: they memorised and have a positive grokking time."""
    return [r for r in recs if r["memorised"] and r["t_grok"] is not None and r["t_grok"] > 0]


recs = records()
fitset = usable(recs)
print(f"{len(fitset)} runs can enter the fits; {sum(not r['memorised'] for r in recs)} never memorised; "
      f"{sum(r['censored'] for r in fitset)} of the usable runs are censored.")
```

## 1. The grid

One row per cell. The medians are over the seeds of the cell. A median of grokking times that includes a
censored seed is itself a lower bound and carries a plus sign. The alignment at the event is the median over
the seeds that generalised, and A_0 is the mean over seeds of the alignment of the initial kernel.

**How to read it.** A run that never memorises has no grokking time and does not enter any fit. It is not
censored, because censoring covers only a missing generalisation event, so the memorised column shows how
many runs each cell loses this way. If A_0 differs between widths, the alignment starts from a different
place at each width, and section 8 accounts for that when it asks whether the threshold is shared.

```python
def med(values, bound):
    """Median of a list of times as a string, with a plus sign when any value is a lower bound."""
    return "-" if not values else f"{np.median(values):.0f}" + ("+" if bound else "")

print(f"{'N':>5} {'eta kappa':>9} {'seeds':>5} {'memorised':>9} {'generalised':>11} {'censored':>8} "
      f"{'t_train':>8} {'t_test':>8} {'t_grok':>8} {'A@event':>8} {'A_0':>7}")
for N in WIDTHS:
    for ek in DECAYS:
        rs = [r for r in recs if (r["N"], r["ek"]) == (N, ek)]
        if not rs:
            continue
        mem = [r for r in rs if r["memorised"]]
        gen = [r for r in mem if not r["censored"]]
        cens = len(mem) - len(gen)
        A_ev = np.median([r["A_ev"] for r in gen]) if gen else None
        print(f"{N:>5} {ek:>9g} {len(rs):>5} {len(mem):>9} {len(gen):>11} {cens:>8} "
              f"{med([r['t_train'] for r in mem], False):>8} {med([r['t_test'] for r in mem], cens > 0):>8} "
              f"{med([r['t_grok'] for r in mem], cens > 0):>8} {L.fmt(A_ev):>8} {np.mean([r['A0'] for r in rs]):>7.4f}")
```

```python
# A_0 by width and seed. The initial kernel does not depend on the decay, so any decay level will do; the
# first saved one is used.
print("Alignment of the initial kernel, A_0, by width and seed")
print(f"{'seed':>5}" + "".join(f"{'N = ' + str(N):>10}" for N in WIDTHS))
for s in SEEDS:
    vals = []
    for N in WIDTHS:
        r = next((r for r in recs if (r["N"], r["seed"]) == (N, s)), None)
        vals.append(f"{r['A0']:>10.4f}" if r else f"{'-':>10}")
    print(f"{s:>5}" + "".join(vals))
for N in WIDTHS:
    a0 = [r["A0"] for r in recs if r["N"] == N and r["ek"] == DECAYS[0]]
    if len(a0) > 1:
        print(f"N = {N}: mean {np.mean(a0):.4f}, standard deviation over seeds {np.std(a0, ddof=1):.4f}")
```

## 2. Learning curves across widths

Train and test accuracy at seed 0 on a log step axis, one colour per width, light for narrow and dark for
wide. The first figure has no decay and the second the strongest decay, 0.0003. Hollow and filled dots are
the memorisation and generalisation events.

**How to read it.** Suppose the memorisation event sits at about the same step at every width while the
generalisation event moves right as N grows. Then the delay grows with width, which is the direction H
predicts. If the generalisation event moves with width at no decay but not at 0.0003, strong decay is
setting the time in those cells, and that is the two-clock case. If the memorisation event also moves with
width, part of the width effect sits in memorisation rather than in the delay, and the t_test column of
section 1 has to be read beside the grokking time. A generalisation event that moves left as N grows is what
the report calls a sign of a parameterisation error, so the first thing to check is then that every run used
the NTK parameterisation.

```python
for ek in [0.0, 3e-4]:
    ws = [N for N in WIDTHS if (N, ek, 0) in grid]
    runs = [grid[(N, ek, 0)] for N in ws]
    fig = L.plot_series(runs, ["train_acc", "test_acc"], [f"N = {N}" for N in ws], [width_colours[N] for N in ws],
                        ["train accuracy", "test accuracy"], legend_panel=1, mark_train=True,
                        crossings=[L.accuracy_crossings(r, TEST_ACC_LEVEL) for r in runs],
                        suptitle=f"Accuracy by width at eta kappa {ek:g}, seed 0, NTK parameterisation")
```

## 3. The kernel terms across widths

The three centred kernel terms at seed 0, one colour per width: the scale term S_t, the shape term R_t and the
centred alignment A_t. The first figure has no decay and the second decay 0.0003.

**How to read it.** H's mechanism is that shape change is a finite-width effect, so a wider network should
turn its kernel more slowly. If R_t and A_t rise more slowly as N grows, and the curves shift right by about
the same factor as the generalisation event, then the width effect on timing runs through the rate of shape
change, as H says. If the shape curves lie on top of each other across widths while the grokking time still
grows, the width effect does not come from shape, and a positive width exponent would then not support H's
account of it. If the scale term also slows with width in step with the event, scale and shape both move with
N here, and the width result on its own cannot tell them apart. That is the same difficulty Tier 1 met under
H2, and it has to be stated.

```python
for ek in [0.0, 3e-4]:
    ws = [N for N in WIDTHS if (N, ek, 0) in grid]
    runs = [grid[(N, ek, 0)] for N in ws]
    fig = L.plot_series(runs, ["S_c", "R_c", "A_t"], [f"N = {N}" for N in ws], [width_colours[N] for N in ws],
                        ["scale S_t", "shape R_t", "alignment A_t"], mark_train=True,
                        crossings=[L.accuracy_crossings(r, TEST_ACC_LEVEL) for r in runs],
                        suptitle=f"Centred kernel terms by width at eta kappa {ek:g}, seed 0, NTK parameterisation")
```

## 4. Grokking time against width and against decay

Left: the grokking time against the width on log axes, one colour per decay level. Right: the grokking time
against the decay on log axes, one colour per width, with the no-decay cells on a separate stub marked 0 so
that they are visible without a made-up log value. Every seed is a point, a circle when measured and a
triangle when it is a lower bound, and the solid lines join the medians of each cell. The dashed lines are
the main-effects fit of section 5, which uses only the nonzero decays.

**How to read it.** Under H the lines in the left panel rise with a common slope and the lines in the right
panel are flat. Under the weight-decay clock of Kim (2026) the right panel falls with slope -1, and the left
panel is flat apart from at most a slow logarithmic drift through the initial weight scale. Kim's Scope
paragraph says the mechanism does not explain grokking without weight decay, so a finite time on the no-decay
stub shows a clock that Kim's theory does not cover rather than refuting it. Lines that rise on the left and
fall on the right mean two clocks. Lines that are not parallel mean an interaction; for example, a steep
width slope at no decay with a flat one at 0.0003 would mean decay is the faster clock only when it is strong.
When most triangles sit at the widest width, the fitted width slope is held down by lower bounds and should
be read as a floor.

```python
def design(rs, interaction=False, centre=None):
    """Design matrix of the log fit: an intercept, log N, log eta kappa, and optionally their product.

    The product is formed from the two logs centred at the given point, so that with the interaction in the
    model the first-order coefficients are still the slopes at the centre of the grid.
    """
    lnN = np.log([r["N"] for r in rs]); lnE = np.log([r["ek"] for r in rs])
    X = np.c_[np.ones(len(rs)), lnN, lnE]
    if interaction:
        cN, cE = centre
        X = np.c_[X, (lnN - cN) * (lnE - cE)]
    return X


def fit_log(rs, response="t_grok", interaction=False, n_boot=1000):
    """Censored fit of log time on log N and log eta kappa over the nonzero-decay runs, resampled within cells."""
    rs = [r for r in rs if r["ek"] > 0]
    centre = (np.mean(np.log(WIDTHS)), np.mean(np.log(NONZERO)))
    X = design(rs, interaction, centre)
    f = L.censored_log_fit(X, [r[response] for r in rs], [r["censored"] for r in rs], n_boot=n_boot, seed=0,
                           groups=[r["cell"] for r in rs])
    f.update(n=len(rs), n_cens=sum(r["censored"] for r in rs), rows=rs, X=X)
    return f


ready = len({r["N"] for r in fitset if r["ek"] > 0}) >= 2 and len({r["ek"] for r in fitset if r["ek"] > 0}) >= 2
main = fit_log(fitset) if ready else None

fig, axes = plt.subplots(1, 2, figsize=(13, 4.6))
ax = axes[0]
for i, ek in enumerate(DECAYS):
    shift = 1.07 ** (i - 2)  # spread the decay levels sideways so the seeds do not overlap
    rs = [r for r in fitset if r["ek"] == ek]
    L.plot_times(ax, [r["N"] * shift for r in rs], [r["t_grok"] for r in rs], [r["censored"] for r in rs],
                 decay_colours[ek])
    ws = [N for N in WIDTHS if any(r["N"] == N for r in rs)]
    ax.plot([N * shift for N in ws], [np.median([r["t_grok"] for r in rs if r["N"] == N]) for N in ws],
            color=decay_colours[ek], lw=1.4, label=f"eta kappa {ek:g}")
    if main is not None and ek > 0:
        c, a, b = main["coef"]
        ax.plot(WIDTHS, np.exp(c + a * np.log(WIDTHS) + b * np.log(ek)), color=decay_colours[ek], lw=1, ls="--")
ax.set_xscale("log"); ax.set_yscale("log")
ax.set_xticks(WIDTHS); ax.set_xticklabels([str(N) for N in WIDTHS]); ax.minorticks_off()
ax.set_xlabel("width N"); ax.set_ylabel("grokking time, steps")
ax.set_title("against width, one line per decay level", loc="left", fontsize=10)
ax.legend(fontsize=7, loc="upper left")
L.tidy_axes(ax); ax.set_xticks(WIDTHS); ax.set_xticklabels([str(N) for N in WIDTHS])

ax = axes[1]
stub = NONZERO[0] / 10 ** 0.8  # where the no-decay cells are drawn
for i, N in enumerate(WIDTHS):
    shift = 1.07 ** (i - 1.5)
    rs = [r for r in fitset if r["N"] == N]
    xs = [(r["ek"] if r["ek"] > 0 else stub) * shift for r in rs]
    L.plot_times(ax, xs, [r["t_grok"] for r in rs], [r["censored"] for r in rs], width_colours[N])
    eks = [ek for ek in NONZERO if any(r["ek"] == ek for r in rs)]
    ax.plot([ek * shift for ek in eks], [np.median([r["t_grok"] for r in rs if r["ek"] == ek]) for ek in eks],
            color=width_colours[N], lw=1.4, label=f"N = {N}")
    zero = [r["t_grok"] for r in rs if r["ek"] == 0]
    if zero:
        ax.plot([stub * shift], [np.median(zero)], "_", color=width_colours[N], ms=16, mew=2)
    if main is not None:
        c, a, b = main["coef"]
        ax.plot(NONZERO, np.exp(c + a * np.log(N) + b * np.log(NONZERO)), color=width_colours[N], lw=1, ls="--")
ax.axvline(stub * 3, color=L.GRAY, lw=0.8, ls=":")
ax.set_xscale("log"); ax.set_yscale("log")
ax.set_xticks([stub] + NONZERO); ax.set_xticklabels(["0"] + [f"{ek:g}" for ek in NONZERO]); ax.minorticks_off()
ax.set_xlabel("eta kappa (no-decay cells on the stub at the left)"); ax.set_ylabel("grokking time, steps")
ax.set_title("against decay, one line per width", loc="left", fontsize=10)
ax.legend(fontsize=7, loc="upper right")
L.tidy_axes(ax)
ax.set_xticks([stub] + NONZERO); ax.set_xticklabels(["0"] + [f"{ek:g}" for ek in NONZERO])
fig.suptitle("Grokking time over the width by decay grid, NTK parameterisation; circles measured, triangles lower bounds, "
             "dashed lines the main-effects fit")
fig.tight_layout()
```

## 5. The (a, b) fit

The head claim. The first table is the main-effects fit log t_grok = c + a log N + b log(eta kappa) over the
runs with nonzero decay. The second adds the product of the two logs, centred at the middle of the grid so
that a and b remain the slopes at that point. The third repeats the main-effects fit on the generalisation
step t_test counted from step 0, which is the quantity Kim (2026), Section 7.5, fits against
(1 - beta)/(eta lambda). All fits are censored regressions with normal noise on the log scale, and the
intervals come from 1,000 bootstrap resamples drawn within each cell. The fourth table fits the width slope
within each decay level, including no decay, and the decay slope within each width, so a single slope is not
assumed.

The figure is the bootstrap cloud of (a, b) for t_grok and for t_test. H predicts a > 0 with b = 0, which is
the dotted line. The weight-decay clock of Kim (2026) at beta = 0, t proportional to 1/(eta lambda), predicts
b = -1 with a at or near 0, the cross. The grey diamond at a = 1 is the value the report attributes to an
O(1/N) finite-width correction. Its source has not been checked, so it is drawn only as a reference.

**How to read it.** A cloud that sits at positive a, straddles b = 0 and excludes (0, -1) supports H. A cloud
near b = -1 with a near 0 supports the weight-decay clock. A cloud away from both axes means two clocks, and
the ratio of the two contributions at each cell says which one is slower. A cloud that covers both
predictions means the design cannot tell them apart, and that has to be reported as it stands; the report
warns about this for a short decay axis, and here the axis has four levels over 1.5 decades. If the t_grok
and t_test clouds disagree on b, the difference comes from subtracting the memorisation step, and Kim's
prediction should be judged against the t_test cloud, because that is the quantity Kim measures. A negative a
is not a result: it points to a parameterisation fault and has to be checked before anything else is
written. If the interval on the product term excludes zero, the main-effects exponents are averages over a
surface that is not a plane, and the within-stratum slopes show where a and b differ.

```python
def show(f, names, title, scale="on the log scale"):
    """Print one fitted model: each coefficient with its interval, then the sample size and residual sigma."""
    print(title)
    for nm, b, lo, hi in zip(names, f["coef"], f["lo"], f["hi"]):
        mark = "   excludes zero" if (lo > 0 or hi < 0) else ""
        print(f"    {nm:28} {b:+.3f}  95% CI [{lo:+.3f}, {hi:+.3f}]{mark}")
    print(f"    runs {f['n']}, censored {f['n_cens']}, residual sigma {np.exp(f['log_sigma']):.3f} {scale}, "
          f"{f['n_boot']} bootstrap fits\n")

if not ready:
    print("Too few widths or decay levels are saved for the fit yet.")
else:
    inter = fit_log(fitset, interaction=True)
    main_test = fit_log(fitset, response="t_test")
    show(main, ["intercept c", "width exponent a", "decay exponent b"],
         "Main effects, log t_grok = c + a log N + b log(eta kappa)")
    show(inter, ["intercept c", "width exponent a", "decay exponent b", "log N x log(eta kappa), centred"],
         "With the interaction")
    show(main_test, ["intercept c", "width exponent a", "decay exponent b"],
         "Main effects on the generalisation step counted from step 0, log t_test")
    if main["coef"][1] < 0:
        print("The width exponent is negative. The report treats this as a sign of a parameterisation fault, so "
              "check that every run used the NTK parameterisation before reading anything else.")
```

```python
def slope(rs, x_of, n_boot=400):
    """Censored fit of log t_grok on one log predictor within a stratum, or None if it cannot be estimated."""
    x = np.log([x_of(r) for r in rs])
    if len(rs) < 4 or len(set(x)) < 2 or all(r["censored"] for r in rs):
        return None
    return L.censored_log_fit(np.c_[np.ones(len(rs)), x], [r["t_grok"] for r in rs], [r["censored"] for r in rs],
                              n_boot=n_boot, seed=0, groups=[r["cell"] for r in rs])

strata = {}
print("Width exponent within each decay level (all widths)")
for ek in DECAYS:
    f = slope([r for r in fitset if r["ek"] == ek], lambda r: r["N"])
    strata[("a", ek)] = f
    if f is not None:
        print(f"    eta kappa {ek:>7g}   a {f['coef'][1]:+.3f}  [{f['lo'][1]:+.3f}, {f['hi'][1]:+.3f}]")
print("Decay exponent within each width (nonzero decays)")
for N in WIDTHS:
    f = slope([r for r in fitset if r["N"] == N and r["ek"] > 0], lambda r: r["ek"])
    strata[("b", N)] = f
    if f is not None:
        print(f"    N = {N:>5}         b {f['coef'][1]:+.3f}  [{f['lo'][1]:+.3f}, {f['hi'][1]:+.3f}]")
```

## 6. Residuals where the power law fails

The residuals of the main-effects fit on the log scale. For a censored run the residual is its expected value
under the fitted model given that the true time exceeds the bound. Left: the mean residual of each cell, with
rows for widths and columns for decay levels. Right: every run's residual against its fitted value, with
censored runs as triangles. The printout names the cells whose mean residual is more than twice its standard
error, sigma over the square root of the number of seeds.

**How to read it.** A heat map without pattern means one power law describes the whole grid. A band of
same-sign residuals at one end of the decay axis means the power law in eta kappa fails there. Examples are a
floor where more decay no longer speeds grokking, or a sharp change where decay takes over. A pattern across
N, such as negative residuals at the middle widths and positive ones at the ends, means the grokking time is
not a power of N. A pattern that follows the diagonal of the grid is the interaction of section 5 seen cell
by cell. The report asks for these cells to be named.

```python
if ready:
    rs, X = main["rows"], main["X"]
    mu = X @ main["coef"]; sigma = np.exp(main["log_sigma"])
    y = np.log([r["t_grok"] for r in rs]); cens = np.array([r["censored"] for r in rs])
    z = (y - mu) / sigma
    resid = np.where(cens, sigma * norm.pdf(z) / norm.sf(z), y - mu)  # expected residual above the bound when censored

    table = np.full((len(WIDTHS), len(NONZERO)), np.nan)
    flagged = []
    for i, N in enumerate(WIDTHS):
        for j, ek in enumerate(NONZERO):
            pick = np.array([(r["N"], r["ek"]) == (N, ek) for r in rs])
            if pick.any():
                table[i, j] = resid[pick].mean()
                if abs(table[i, j]) > 2 * sigma / np.sqrt(pick.sum()):
                    flagged.append((N, ek, table[i, j], pick.sum()))

    fig, axes = plt.subplots(1, 2, figsize=(12, 4.2))
    ax = axes[0]
    span = np.nanmax(np.abs(table))
    im = ax.imshow(table, cmap="RdBu_r", vmin=-span, vmax=span)
    for i in range(len(WIDTHS)):
        for j in range(len(NONZERO)):
            if not np.isnan(table[i, j]):
                ax.text(j, i, f"{table[i, j]:+.2f}", ha="center", va="center", fontsize=8)
    ax.set_xticks(range(len(NONZERO))); ax.set_xticklabels([f"{ek:g}" for ek in NONZERO])
    ax.set_yticks(range(len(WIDTHS))); ax.set_yticklabels([str(N) for N in WIDTHS])
    ax.set_xlabel("eta kappa"); ax.set_ylabel("width N"); ax.grid(False)
    ax.set_title("mean residual per cell, log scale", loc="left", fontsize=10)
    fig.colorbar(im, ax=ax, fraction=0.04, pad=0.02).set_label("observed minus fitted log time")
    ax = axes[1]
    for N in WIDTHS:
        pick = np.array([r["N"] == N for r in rs])
        L.plot_times(ax, mu[pick], resid[pick], cens[pick], width_colours[N], label=f"N = {N}")
    ax.axhline(0, color=L.GRAY, lw=1)
    ax.set_xlabel("fitted log grokking time"); ax.set_ylabel("residual")
    ax.set_title("residual against fitted value", loc="left", fontsize=10)
    ax.legend(fontsize=7)
    L.tidy_axes(ax)
    fig.suptitle("Where the main-effects power law fits and where it does not")
    fig.tight_layout()

    print(f"Residual sigma {sigma:.3f}. Cells whose mean residual exceeds twice its standard error:")
    for N, ek, v, n in flagged:
        print(f"    N = {N:>5}, eta kappa {ek:g}: mean residual {v:+.3f} over {n} runs")
    if not flagged:
        print("    none")
```

## 7. The no-decay column and the reciprocal fit

The log fit cannot use the no-decay cells. The report's reciprocal form
1/t_grok = c0 + c1 eta kappa + c2 log N keeps them, because eta kappa = 0 is simply the intercept. Here the
rate is written per 10,000 steps and the decay in units of 0.0001, so that the coefficients are of order
one. The model is fitted twice with the same likelihood. The first fit uses log N as printed, so its
intercept c0 is the no-decay rate at N = 1, which lies outside the grid. The second fit uses log N minus
log 100, so its intercept c0' is the no-decay rate at the baseline width. The noise is normal on the rate
scale. A censored run enters as an upper bound on its rate, because a lower bound on a time is an upper bound
on its reciprocal. Both choices are ours; the report does not specify them.

The figure shows the rate against the decay on linear axes that include zero, one colour per width, with
censored runs as downward triangles and the fitted line of each width.

**How to read it.** Under a pure weight-decay clock every line passes through the origin. That needs c0 and
c2 both zero, not c0 alone, and the printout gives the joint test. A positive no-decay rate at a width means
grokking runs on a clock that does not need decay. Kim's Scope paragraph says the weight-decay mechanism does
not account for such a clock. The share column bounds how much of the timing decay accounts for,
independently of b. If the no-decay rates fall towards zero as N grows, the no-decay clock slows with width,
which is H's width effect seen from the zero-decay side. If the log fit and the reciprocal fit disagree, for
example b near 0 but a large decay share, the report treats that as diagnostic, and the usual cause is
curvature in the decay axis that section 6 should show.

```python
RATE, UNIT = 1e4, 1e-4  # rate per 10,000 steps; decay in units of 0.0001

def fit_reciprocal(rs, centred, n_boot=1000):
    """Censored fit of the rate on the decay and log N, with censored runs as upper bounds on the rate."""
    lnN = np.log([r["N"] for r in rs]) - (np.log(100) if centred else 0.0)
    X = np.c_[np.ones(len(rs)), [r["ek"] / UNIT for r in rs], lnN]
    rate = RATE / np.array([r["t_grok"] for r in rs], float)
    f = L.censored_linear_fit(X, rate, [r["censored"] for r in rs], side="left", n_boot=n_boot, seed=0,
                              groups=[r["cell"] for r in rs])
    f.update(n=len(rs), n_cens=sum(r["censored"] for r in rs))
    return f

if len({r["N"] for r in fitset}) >= 2:
    rec_raw = fit_reciprocal(fitset, centred=False)
    rec = fit_reciprocal(fitset, centred=True)
    show(rec_raw, ["c0, rate at N = 1", "c1, per 0.0001 of decay", "c2, per unit of log N"],
         "Reciprocal form, rate per 10,000 steps = c0 + c1 (eta kappa / 0.0001) + c2 log N", "in rate units")
    show(rec, ["c0', rate at N = 100", "c1, per 0.0001 of decay", "c2, per unit of log N"],
         "The same model with log N centred at the baseline width", "in rate units")

    # Joint test of c0 = c2 = 0 with the bootstrap covariance of the two coefficients.
    v = rec_raw["coef"][[0, 2]]; cov = np.cov(rec_raw["boot"][:, [0, 2]].T)
    stat = float(v @ np.linalg.solve(cov, v))
    print(f"Joint test that c0 = c2 = 0, which a pure weight-decay clock requires: Wald statistic {stat:.1f}, "
          f"p = {chi2.sf(stat, 2):.2g}\n")

    print(f"{'N':>6} {'no-decay rate':>28}")
    for N in WIDTHS:
        draws = rec["boot"][:, 0] + rec["boot"][:, 2] * np.log(N / 100)
        point = rec["coef"][0] + rec["coef"][2] * np.log(N / 100)
        lo, hi = np.percentile(draws, [2.5, 97.5])
        print(f"{N:>6} {point:>10.3f} [{lo:+.3f}, {hi:+.3f}]")

    print(f"\nShare of the fitted rate at N = 100 that comes from decay")
    for ek in NONZERO:
        x = ek / UNIT
        share = rec["boot"][:, 1] * x / (rec["boot"][:, 0] + rec["boot"][:, 1] * x)
        point = rec["coef"][1] * x / (rec["coef"][0] + rec["coef"][1] * x)
        lo, hi = np.percentile(share, [2.5, 97.5])
        print(f"    eta kappa {ek:>7g}: {point:.0%}  [{lo:.0%}, {hi:.0%}]")

    fig, ax = plt.subplots(figsize=(7.5, 4.6))
    for i, N in enumerate(WIDTHS):
        rs = [r for r in fitset if r["N"] == N]
        if not rs:
            continue
        jitter = (i - 1.5) * 0.03
        L.plot_times(ax, [r["ek"] / UNIT + jitter for r in rs], [RATE / r["t_grok"] for r in rs],
                     [r["censored"] for r in rs], width_colours[N], flip=True, label=f"N = {N}")
        xs = np.linspace(0, max(DECAYS) / UNIT, 50)
        c0, c1, c2 = rec["coef"]
        ax.plot(xs, c0 + c1 * xs + c2 * np.log(N / 100), color=width_colours[N], lw=1.2, ls="--")
    ax.axhline(0, color=L.GRAY, lw=0.8); ax.axvline(0, color=L.GRAY, lw=0.8)
    ax.set_xlabel("eta kappa, in units of 0.0001"); ax.set_ylabel("grokking rate, per 10,000 steps")
    ax.set_title("Reciprocal of the grokking time against decay; a pure decay clock passes through the origin",
                 loc="left", fontsize=10)
    ax.legend(fontsize=8, loc="upper left")
    L.tidy_axes(ax)
    fig.tight_layout()
```

## 8. Is the threshold A_* shared across widths?

This section follows section 6 of the Tier 1 notebook. Each panel reads one kernel term at the
generalisation event for every run that generalised, grouped by width and coloured by decay. The panels are
the alignment A, the distance A minus A_0 that the alignment travelled, the shape term R, and the kernel norm
ratio exp(S). The band is the mean plus or minus one standard deviation over runs. The title gives the
max-to-min ratio and the spread, the standard deviation as a share of the mean. The second figure repeats the
Tier 1 placebo test on these runs: each run is read at another run's generalisation step, 2,000 times, and
the spread at the true events is compared with the spread at the swapped ones.

**How to read it.** H says A_* is shared across N. A max-to-min ratio under about two on the alignment panel,
with the swapped spread wider than the true spread, extends the Tier 1 threshold to the width axis. A_0 rises
with N, so if A_* stays flat the distance to the threshold shrinks with width. That shortens the grokking
time at large N and works against a positive width exponent, so a positive exponent measured anyway is
conservative. If A_* itself rises with N, the threshold depends on width, and H1 fails on this axis even
though it held across alpha and decay. Tier 1 found that the task-blind shape term R is locked to the event
more tightly than A is. If that happens again here it has to be reported beside the alignment result.

```python
gen = [r for r in fitset if not r["censored"]]
panels = [("alignment A", lambda r: r["A_ev"], False), ("distance A - A_0", lambda r: r["A_ev"] - r["A0"], False),
          ("shape R", lambda r: r["R_ev"], False), ("kernel norm ratio exp(S)", lambda r: np.exp(r["S_ev"]), True)]
offsets = dict(zip(DECAYS, np.linspace(-0.15, 0.15, len(DECAYS))))

def spread(v):
    """The max-to-min ratio and the standard deviation as a share of the mean."""
    v = np.asarray(v, float)
    return v.max() / v.min(), v.std(ddof=1) / v.mean()

if len(gen) > 2:
    fig, axes = plt.subplots(1, 4, figsize=(18, 4))
    for ax, (label, get, logy) in zip(axes, panels):
        vals = np.array([get(r) for r in gen])
        for r, v in zip(gen, vals):
            ax.plot([WIDTHS.index(r["N"]) + offsets[r["ek"]]], [v], "o", color=decay_colours[r["ek"]], ms=5,
                    alpha=0.85, markeredgecolor="white", markeredgewidth=0.5)
        ax.axhspan(vals.mean() - vals.std(ddof=1), vals.mean() + vals.std(ddof=1), color=L.GRAY, alpha=0.12)
        ax.axhline(vals.mean(), color=L.GRAY, lw=1, ls="-.")
        ratio, cv = spread(vals)
        ax.set_title(f"{label}: max/min {ratio:.2f}, spread {cv:.0%}", loc="left", fontsize=10)
        ax.set_xticks(range(len(WIDTHS))); ax.set_xticklabels([f"N = {N}" for N in WIDTHS])
        if logy:
            ax.set_yscale("log")
        L.tidy_axes(ax)
    for ek in DECAYS:
        axes[0].plot([], [], "o", color=decay_colours[ek], label=f"eta kappa {ek:g}")
    axes[0].legend(fontsize=7, loc="best", ncol=2)
    fig.suptitle(f"Kernel terms at the generalisation event across widths, {len(gen)} runs, NTK parameterisation")
    fig.tight_layout()

    print(f"{'N':>6} {'runs':>5} {'A at event':>11} {'A_0':>8} {'distance':>9} {'R at event':>11}")
    for N in WIDTHS:
        rs = [r for r in gen if r["N"] == N]
        if rs:
            print(f"{N:>6} {len(rs):>5} {np.mean([r['A_ev'] for r in rs]):>11.4f} {np.mean([r['A0'] for r in rs]):>8.4f} "
                  f"{np.mean([r['A_ev'] - r['A0'] for r in rs]):>9.4f} {np.mean([r['R_ev'] for r in rs]):>11.4f}")
```

```python
# The placebo: every run read at another run's generalisation step, two thousand times.
TERMS = [("A", "A_t", lambda x: x), ("R", "R_c", lambda x: x), ("exp(S)", "S_c", np.exp)]
if len(gen) > 2:
    gen_runs = [grid[(r["N"], r["ek"], r["seed"])] for r in gen]
    steps = np.array([r["t_test"] for r in gen])
    observed = {nm: spread([tr(L.value_at_step(run, key, t)) for run, t in zip(gen_runs, steps)])[1]
                for nm, key, tr in TERMS}
    rng = np.random.default_rng(0)
    draws = {nm: [] for nm, _, _ in TERMS}
    for _ in range(2000):
        perm = rng.permutation(len(gen))
        while np.any(perm == np.arange(len(gen))):  # no run may keep its own step
            perm = rng.permutation(len(gen))
        for nm, key, tr in TERMS:
            draws[nm].append(spread([tr(L.value_at_step(gen_runs[i], key, steps[perm[i]])) for i in range(len(gen))])[1])

    print(f"{'term':>8} {'spread at its own event':>24} {'median at a swapped event':>26} {'tightening':>11} {'p':>7}")
    for nm, _, _ in TERMS:
        d = np.array(draws[nm]); o = observed[nm]
        print(f"{nm:>8} {o:>23.1%} {np.median(d):>25.1%} {np.median(d) / o:>10.1f}x {np.mean(d <= o):>7.3f}")

    fig, axes = plt.subplots(1, 3, figsize=(14, 3.8))
    for ax, (nm, _, _) in zip(axes, TERMS):
        d = np.array(draws[nm])
        ax.hist(d, bins=40, color=L.RAMP[1], edgecolor="white", lw=0.4)
        ax.axvline(observed[nm], color=L.ORANGE, lw=2)
        ax.set_xlabel(f"spread of {nm} across runs"); ax.set_ylabel("permutations")
        ax.set_title(f"{nm}: own event {observed[nm]:.1%}, swapped {np.median(d):.1%}", loc="left", fontsize=10)
        L.tidy_axes(ax)
    fig.suptitle("Placebo test on the Tier 2 runs: the orange line is the spread at the true generalisation events")
    fig.tight_layout()
```

## 9. Distance over rate: does the early shape rate predict the grokking time?

This is a mechanism check that goes beyond the report's list for Tier 2. The proposal argues that under H
the grokking time is a distance divided by a rate, and that a wider network turns its kernel more slowly.
For each run the early rate of the alignment and of the shape term is the least-squares slope over steps 0
to 2,000. That window was fixed before any Tier 2 run was seen, and the first printout checks whether it ends
before memorisation in every cell. The figure plots the grokking time against each early rate on log axes,
coloured by width, with one marker shape per decay level. Hollow markers are censored runs. The printout
gives the censored slope of log grokking time on log early rate, alone and with log N added.

**How to read it.** If the early rate falls with width and the points lie on one line of slope near -1 across
both widths and decays, the time is set by how fast the kernel turns, which is H's mechanism. It is also the
early-warning claim in the proposal's significance section. If log N adds nothing once the rate is in the
model, the width effect of section 5 passes through the rate. If decayed cells sit below the line, grokking
before their rate predicts, decay provides a second route, which agrees with a two-clock reading of
section 5. If the early rate does not change with width, the width effect of section 5 does not pass through
the rate of shape change.

```python
WINDOW = 2000  # fixed before any Tier 2 run was seen

def early_rate(run, key):
    """Least-squares slope of a history key over the checkpoints from step 0 to WINDOW, per 1,000 steps."""
    h = run["history"]
    steps = np.asarray(h["step"]); pick = steps <= WINDOW
    return np.polyfit(steps[pick] / 1000, np.asarray(h[key])[pick], 1)[0]

for r in fitset:
    run = grid[(r["N"], r["ek"], r["seed"])]
    r["rate_A"], r["rate_R"] = early_rate(run, "A_t"), early_rate(run, "R_c")
early = [r for r in fitset if r["t_train"] <= WINDOW]
print(f"Runs that memorised inside the window: {len(early)} of {len(fitset)}"
      + ("" if not early else f", in cells {sorted({r['cell'] for r in early})}"))

markers = dict(zip(DECAYS, ["o", "s", "D", "P", "X"]))
fig, axes = plt.subplots(1, 2, figsize=(13, 4.6))
for ax, key, label in [(axes[0], "rate_A", "early rate of the alignment A"), (axes[1], "rate_R", "early rate of the shape term R")]:
    rs = [r for r in fitset if r[key] > 0]
    for r in rs:
        colour = width_colours[r["N"]]
        ax.plot([r[key]], [r["t_grok"]], markers[r["ek"]], ms=6, alpha=0.85, markeredgecolor=colour,
                markerfacecolor="none" if r["censored"] else colour, markeredgewidth=1.2)
    ax.set_xscale("log"); ax.set_yscale("log")
    ax.set_xlabel(f"{label}, per 1,000 steps over steps 0 to {WINDOW}"); ax.set_ylabel("grokking time, steps")
    dropped = len(fitset) - len(rs)
    x = np.log([r[key] for r in rs])
    title = f"{label}"
    if len(rs) > 5:
        f1 = L.censored_log_fit(np.c_[np.ones(len(rs)), x], [r["t_grok"] for r in rs], [r["censored"] for r in rs],
                                n_boot=400, seed=0, groups=[r["cell"] for r in rs])
        f2 = L.censored_log_fit(np.c_[np.ones(len(rs)), x, np.log([r["N"] for r in rs])], [r["t_grok"] for r in rs],
                                [r["censored"] for r in rs], n_boot=400, seed=0, groups=[r["cell"] for r in rs])
        xs = np.linspace(x.min(), x.max(), 2)
        ax.plot(np.exp(xs), np.exp(f1["coef"][0] + f1["coef"][1] * xs), color=L.GRAY, lw=1.2, ls="--")
        title += f": slope {f1['coef'][1]:+.2f}"
        print(f"{label}: {len(rs)} runs, {dropped} dropped with a rate at or below zero")
        print(f"    alone          slope {f1['coef'][1]:+.3f} [{f1['lo'][1]:+.3f}, {f1['hi'][1]:+.3f}], "
              f"residual sigma {np.exp(f1['log_sigma']):.3f}")
        print(f"    with log N     slope {f2['coef'][1]:+.3f} [{f2['lo'][1]:+.3f}, {f2['hi'][1]:+.3f}], "
              f"log N {f2['coef'][2]:+.3f} [{f2['lo'][2]:+.3f}, {f2['hi'][2]:+.3f}]")
    ax.set_title(title, loc="left", fontsize=10)
    L.tidy_axes(ax)
for N in WIDTHS:
    axes[0].plot([], [], "o", color=width_colours[N], label=f"N = {N}")
for ek in DECAYS:
    axes[0].plot([], [], markers[ek], color=L.GRAY, label=f"eta kappa {ek:g}")
axes[0].plot([], [], "o", markerfacecolor="none", markeredgecolor=L.GRAY, label="censored")
axes[0].legend(fontsize=7, loc="best", ncol=2)
fig.suptitle("Grokking time against the early rate of shape change; colour is width, marker is decay")
fig.tight_layout()
```

## 10. Sensitivity to the definition of the events

The main-effects log fit and the reciprocal intercept at the baseline width, refitted with the generalisation
event at test accuracy 1.0, 0.95 and 0.9, and with the loss definition at a threshold of 0.01 on both losses.
The loss definition is expected to be almost entirely censored, as in Tier 0, because the test loss settles
between 0.01 and 0.03.

**How to read it.** If the signs and rough sizes of a and b do not change across definitions, the head claim
does not depend on the definition of the events. If they change, the claim has to be stated for the primary
definition only, and the change has to be reported.
